# Hyperparameter Search

A structured hyperparameter search is essential to obtain models with strong generalization (i.e., low holdout neuron loss) rather than just good training fit.

The recommended pipeline for MR-LFADS is as follows:

## 1. Fixed parameters: 

Architectural choices (e.g., generator size, controller size) typically have second-order impact on fit quality, provided they are not underparameterized. The goal is to avoid capacity bottlenecks without over-tuning architecture.

* GRU hidden size: Use $~2x$ the number of neurons in the corresponding area. If an area has very few neurons (e.g., $<20$), consider excluding it from modeling.
* Inferred input / communication channel dimensionality: Set to a moderately large value (e.g., 6–12). Overparameterization is acceptable here because KL regularization will suppress unused dimensions.

## 2. Per-Area Grid Search on KL(u)

Next, optimize the inferred input KL coefficient independently for each area. For each area, train single-region LFADS (SR-LFADS) model and perform a grid search over KL(u) values. Select the value that minimizes holdout neuron loss.

## 3. Joint Grid Search on KL(m) and Random Seed

With all areas combined, and KL(u) fixed per area, perform a grid search over KL(m) and random seed values. A rule of thumb is to set KL(m) to be 1–2 orders of magnitude smaller than KL(u).

An example script for grid search using `ray tune` is provided in `examples/01_memory_network/hpsearch.py`.
